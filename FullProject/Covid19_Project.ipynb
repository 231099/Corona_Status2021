<h2>First Import Pandas Library And Assign Dataset Into Dataframe </h2>


```python
import pandas as pd

df = pd.read_csv(r"C:\Users\user\Downloads\COVID19_Statistics_200_Rows-1.csv")

df
```

<h2>Second Split Date Column into Year, Month, Day </h2>


```python
# First Change Date Datatype from object to date

df['Date'] = pd.to_datetime(df['Date'])

# Then split the date column 

df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day


print(df['Date'].dtype)   # should say datetime64[ns]
print(df['Month'].dtype)  # should say int32

```

```python
#Change Month Column into Text instead of regular number

month_map = {
    1: 'January', 2: 'February', 3: 'March', 4: 'April',
    5: 'May', 6: 'June', 7: 'July', 8: 'August',
    9: 'September', 10: 'October', 11: 'November', 12: 'December'
}
df['Month_Name'] = df['Month'].map(month_map)

df
```

```python
#The raw dataset was cleaned by removing rows where the Active column was 0 because these records violated the relationship: Confirmed = Active + Recovered + Deaths. Since the values were inconsistent, they were excluded to ensure the accuracy and reliability of the analysis.
#Assign the new cleaned dataset into new dataframe then save it

df_1 = df.drop(df[df["Active"]==0].index)

df_1.to_csv(r"C:\Users\user\Downloads\COVID19_Statistics_Cleaned.csv", index=False)
```
